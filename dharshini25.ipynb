```python
import nltk
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
texts = " hello im working in python for my lab work"
for text in texts:
    snt= nltk.sent_tokenize(text)
    for s in snt:
        w=nltk.word_tokenize(s)
        tw=nltk.pos_tag(w)
        print(tw)
```
