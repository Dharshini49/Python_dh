# Class Inheritance

```python
class mypar: # parent class
     def greet(self):
         print("Hello, parent class")

class mych(mypar): # child class
    pass
obj= mych()
obj.greet() # accessing parent's class variable with child class.
```

```python
class mypar: # parent class
     def greet(self):
         print("Hello, parent class")
# child class
class mych(mypar): 
    def greet(self):
        super().greet() # calls function from the parent class 
        print("Hello, child class")
obj= mych()
obj.greet()
```

```python
class myemp(emp):
    pass
ob=myemp('karunya','lakshmi',90000)
print(ob.pay)
ob.apply_raise()
print(ob.pay)
```

```python
# Multiple inheretance
class A:
    def func_a(self):
        print("I'm A")
class B:
    def func_b(self):
        print("I'm B")
class C(A,B): # inheritance from both A and B
    pass
obj=C()
obj.func_a()
obj.func_b()
```

```python
class contact:
    all_contacts=[]
    def __init__(self,name,email):
        self.name=name
        self.email=email
        contact.all_contacts.append(self)
class supplier(contact):
    def order(self, order):
        print("Send" "{} order to {}".format(order,self.name))
```

```python
c = contact("some body","somebody@gamil.com")
s = supplier("sup plier","supplier@gmail.com")
print(c.name,c.email,s.name,s.email)
```
