# EXCEL

```python
# Reading an EXCEL file.
import pandas as pd
data= pd.read_excel("/home/matlab/Downloads/Financial Sample.xlsx")
```

```python
# extracting data from the excel file
data
```

```python
# Query -- given the condition that print all the employee details whos country is Mexico.
```

```python
res = data[data['Country'] == 'Mexico']
```

```python
# printing only the empolyee with specific condition 
```

```python
res
```

```python
# writing the query result to a new Excel file
```

```python
op_file= "Mexico_emp.xlsx"
res.to_excel(op_file, index=False)
```

```python
op_file
```

```python
# new file based on the query is created successfully
```

# JSON

```python
# opening this output file in JSON format 
```

```python
df = pd.read_excel(op_file)
```

```python
df
```

```python
json_op = df.to_json(orient="records",indent=4)
```

```python
op_json = "mex_emp.json"
```

```python
# save to a JSON file
with open(op_json, "w") as f:
    f.write(json_op)
```
