```python
# 1. Passing Arguments - Keyword Arguments

def fullname(fn,ln):
    full_name=f"{fn} {ln}"
    return full_name.title()
print(fullname(fn='dharshini',ln='r'))
```

```python
# 2. Multiple Function Calls
# Passing Arguments - Positional Arguments
fullname('karunya','lakshmi') # directly calling a function without print function 
```

```python
# Passing Arguments - Keyword Arguments
fullname('anu','keerthana')
```

```python
# 3. order matters in positional arguments
def describe_pet(atype,name):
    print(f"\nI have a {atype}.")
    print(f"My {atype}'s name is {name.title()}.")
```

```python
describe_pet('cat','pussy')
```

```python
describe_pet('mujju','dog') #example for order displacement
```

```python
# 4. Keyword Arguments (order doesn’t matter)
describe_pet(name='harry', atype='hamster')
```

```python
# 5. Default Values
def describe_pet(atype,name='bruno'):
    print(f"\nI have a {atype}.")
    print(f"My {atype}'s name is {name.title()}.")

describe_pet('cat')
```

```python
# 6. Many ways to call a function (commented examples)
# describe_pet('bruno') --- default type is dog
# describe_pet(name='bruno') --- same as above
# describe_pet('harry','hamster') --- both given
# describe_pet(name='harry',atype='hamster') --- keyword args
# describe_pet(atype='hamster',name='harry') --- keyword args
```

```python
# 7. Returning a simple value
def full_name(fn, ln):
    return f"{fn.title()} {ln.title()}"
print("Full Name:",full_name('dharshini','r'))
```

```python
# 8. Making an Argument Optional
def _fullname(fn, ln, middle=''):
    if middle:
        return f"{fn.title()} {middle.title()} {ln.title()}"
    else:
        return f"{fn.title()} {ln.title()}"

print(_fullname('dharshini', 'r'))
print(_fullname('mythili', 'keerthana', 'anu'))
```

```python
# 9. Returning a Dictionary --- arbitrary keywords in function
def userinfo(fn, ln, **info):
    info['fname']=fn
    info['lname']=ln
    return info
profile= userinfo('dharshini','r',age=20,branch='MID',year=4)
print(profile)
```

```python
# 10. Passing a List
def greet(names):
    for name in names:
        print(f"Hello, {name.title()}!")
greet(['dharshini','agnes','varshitha'])
```

```python
# 11. Modifying a List inside the function (HW)
def wish(friends):
    for friend in range(len(friends)):
        friends[friend] = "Good Morning " + friends[friend]
friends = ['agnes', 'varshitha']
wish(friends)
print(friends)  # list changed
```

```python
# 12. Protecting a List from Updates (HW)
friends = ['agnes', 'varshitha']
wish(friends[:])  # pass a copy
print(friends)  # original unchanged
```

```python
# 13. Passing an Arbitrary Number of Arguments
def make_pizza(*toppings):
    for topping in toppings:
        print(f"{topping} pizza")
make_pizza('pepperoni', 'mushrooms', 'cheese')
```

```python
# 14. Mixing Positional and Arbitrary Arguments
def make_sandwich(size, *fillings):
    print(f"\n{size}-inch sandwich with:")
    for filling in fillings:
        print(f"- {filling}")

make_sandwich(12, 'chicken', 'lettuce', 'mayo')
```
