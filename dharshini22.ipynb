```python
#basic user inputs
def info():
    name = input("Enter your name: ")
    dest = input("Enter your destination: ")
    sdate = input("Travel start date: ")
    edate = input("Travel end date: ")
    travelers = int(input("Number of travelers: "))
    print("\nPackage Types:\n1. Budget\n2. Standard\n3. Luxury")
    choice = input("Choose a package (1/2/3): ")
    packages = {'1': 'Budget', '2': 'Standard', '3': 'Luxury'}
    package = packages.get(choice, 'Standard')
    
    return {
        'name': name,
        'destination': dest,
        'start_date': sdate,
        'end_date': edate,
        'travelers': travelers,
        'package': package
    }
```

```python
def calcost(travelers, package):
    base_prices = {'Budget': 10000, 'Standard': 20000, 'Luxury': 35000}
    return base_prices[package] * travelers
```

```python
def dashboard(info):
    cost = calcost(info['travelers'], info['package'])
    print(f"Name: {info['name']}")
    print(f"Destination: {info['destination']}")
    print(f"Travel Dates: {info['start_date']} to {info['end_date']}")
    print(f"Travelers: {info['travelers']}")
    print(f"Package Chosen: {info['package']}")
    print(f"Estimated Total Cost: ₹{cost}")
    print("===================================")
```

```python
user_info = info()
```

```python
dashboard(user_info)
```
